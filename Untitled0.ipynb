```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

def scrape_quotes(jumlah_halaman):
    data = []
    for page in range(1, jumlah_halaman + 1):
        url = f"https://quotes.toscrape.com/page/{page}/"
        response = requests.get(url)
        soup = BeautifulSoup(response.text, "html.parser")
        containers = soup.find_all("div", class_="quote")
        for c in containers:
            text = c.find("span", class_="text").text
            author = c.find("small", class_="author").text
            data.append({"Page": page, "Author": author, "Quote": text})
    return pd.DataFrame(data)

df = scrape_quotes(3)
top_author = df["Author"].value_counts().idxmax()
top_count = df["Author"].value_counts().max()

summary = pd.DataFrame({
    "Metrik": ["Total Quote Dianalisis", "Penulis Paling Sering", "Jumlah Kemunculan"],
    "Nilai": [len(df), top_author, top_count]
})

with pd.ExcelWriter("laporan_quotes.xlsx") as writer:
    df.to_excel(writer, sheet_name="Data Lengkap", index=False)
    summary.to_excel(writer, sheet_name="Ringkasan", index=False)

print("Laporan tersimpan!")
```
